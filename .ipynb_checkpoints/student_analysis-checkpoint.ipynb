# Attributes for both student-mat.csv (Math course) and student-por.csv (Portuguese language course) datasets:
    1 school - student's school (binary: "GP" - Gabriel Pereira or "MS" - Mousinho da Silveira)
    2 sex - student's sex (binary: "F" - female or "M" - male)
    3 age - student's age (numeric: from 15 to 22)
    4 address - student's home address type (binary: "U" - urban or "R" - rural)
    5 famsize - family size (binary: "LE3" - less or equal to 3 or "GT3" - greater than 3)
    6 Pstatus - parent's cohabitation status (binary: "T" - living together or "A" - apart)
    7 Medu - mother's education (numeric: 0 - none,  1 - primary education (4th grade), 2 – 5th to 9th grade, 3 – secondary education or 4 – higher education)
    8 Fedu - father's education (numeric: 0 - none,  1 - primary education (4th grade), 2 – 5th to 9th grade, 3 – secondary education or 4 – higher education)
    9 Mjob - mother's job (nominal: "teacher", "health" care related, civil "services" (e.g. administrative or police), "at_home" or "other")
    10 Fjob - father's job (nominal: "teacher", "health" care related, civil "services" (e.g. administrative or police), "at_home" or "other")
    11 reason - reason to choose this school (nominal: close to "home", school "reputation", "course" preference or "other")
    12 guardian - student's guardian (nominal: "mother", "father" or "other")
    13 traveltime - home to school travel time (numeric: 1 - <15 min., 2 - 15 to 30 min., 3 - 30 min. to 1 hour, or 4 - >1 hour)
    14 studytime - weekly study time (numeric: 1 - <2 hours, 2 - 2 to 5 hours, 3 - 5 to 10 hours, or 4 - >10 hours)
    15 failures - number of past class failures (numeric: n if 1<=n<3, else 4)
    16 schoolsup - extra educational support (binary: yes or no)
    17 famsup - family educational support (binary: yes or no)
    18 paid - extra paid classes within the course subject (Math or Portuguese) (binary: yes or no)
    19 activities - extra-curricular activities (binary: yes or no)
    20 nursery - attended nursery school (binary: yes or no)
    21 higher - wants to take higher education (binary: yes or no)
    22 internet - Internet access at home (binary: yes or no)
    23 romantic - with a romantic relationship (binary: yes or no)
    24 famrel - quality of family relationships (numeric: from 1 - very bad to 5 - excellent)
    25 freetime - free time after school (numeric: from 1 - very low to 5 - very high)
    26 goout - going out with friends (numeric: from 1 - very low to 5 - very high)
    27 Dalc - workday alcohol consumption (numeric: from 1 - very low to 5 - very high)
    28 Walc - weekend alcohol consumption (numeric: from 1 - very low to 5 - very high)
    29 health - current health status (numeric: from 1 - very bad to 5 - very good)
    30 absences - number of school absences (numeric: from 0 to 93)
    
    # these grades are related with the course subject, Math or Portuguese:
    31 G1 - first period grade (numeric: from 0 to 20)
    31 G2 - second period grade (numeric: from 0 to 20)
    32 G3 - final grade (numeric: from 0 to 20, output target)
    
    Additional note: there are several (382) students that belong to both datasets . 
    These students can be identified by searching for identical attributes
    that characterize each student, as shown in the annexed R file.

## Is study time correlated with performance?


```python
metrics= df[['studytime', 'G1', 'G2', 'G3']].corr()
```

```python
plt.figure(figsize=(6,6))
sns.heatmap(metrics, annot=True, cmap='coolwarm')
```

* Based on correlation Heatmap we can say that there is weak positive correlation betwenn study time & marks (G1, G2, G3)
* But heatmap also showing that G1,G2,G3 are highly correlated with each other that mean the students performance is depend on past marks

## Which gender performs better on average?

```python
df['marks']=df['G1'] + df['G2'] + df['G3']
```

```python
df['marks'].head()
```

```python
df.head()
```

Now we created a new column named marks which has total marks of students

lets see students performance based on gender

```python
plt.figure(figsize=(6,6))
female_avg= df[df['sex']=='F']['marks'].mean()
male_avg= df[df['sex']=='M']['marks'].mean()
print(male_avg)
sns.histplot(data=df, x='marks', hue='sex', bins=5)   #, multiple='dodge'
plt.axvline(female_avg, color='blue', linestyle='--', linewidth=2, label=f'Female Avg: {female_avg:.2f}')
plt.axvline(male_avg, color='orange', linestyle='--', linewidth=2, label=f'Male Avg: {male_avg:.2f}')

plt.title("Students Performance Based on Gender")
plt.xlabel("Marks")
plt.ylabel("Count")
plt.legend()
plt.show()
```

```python
fig, axes= plt.subplots(1,2, figsize=(14,6))

df_male= df[ df['sex'] == 'M']
df_female= df[ df['sex'] == 'F']

#1st plot
sns.histplot(ax= axes[0], data=df_female, x= 'marks', color= 'pink')
female_avg = df_female['marks'].mean()
axes[0].axvline(female_avg, color='darkblue', linestyle='--', linewidth=2, label=f'Avg: {female_avg:.2f}')
axes[0].set_title('Female Student Performance')
axes[0].set_xlabel('Marks')
axes[0].set_ylabel('Count')
axes[0].legend()


#2nd plot
sns.histplot(ax= axes[1], data= df_male, x= 'marks', color= 'blue')
male_avg = df_male['marks'].mean()
axes[1].axvline(male_avg, color='pink', linestyle='--', linewidth=2, label=f'Avg: {male_avg:.2f}')
axes[1].set_title('Male Student Performance')
axes[1].set_xlabel('Marks')
axes[1].set_ylabel('Count')
axes[1].legend()
```

* The above graphs explains us how the Students performance based on Gendeer
* male student has avg marks of 33.22
* female students has avg marks of 30.98 ~ 31

# 4. Visualizations
• Histogram of grades.
• Scatterplot: study time vs grades.
• Bar chart: male vs female average score.

## Histogram of grades

```python
fig, axes = plt.subplots(1,3, figsize=(21,6))
sns.histplot(ax= axes[0], data=df, x='G1', color='orange', bins=5)
axes[0].set_title("Distribution of First period grade (G1)")

sns.histplot(ax= axes[1], data=df, x='G2', color='blue', bins=5)
axes[1].set_title("Distribution of Second period grade (G2)")

sns.histplot(ax= axes[2], data=df, x='G3', color='green', bins=5)
axes[2].set_title("Distribution of Final Grades (G3)")

plt.show()
```

## Scatterplot: study time vs grades.

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# Plot 1: Study time vs G1
axes[0].scatter(df['studytime'], df['G1'], color='blue', alpha=0.5)
axes[0].set_title("Distribution of First period grade (G1)")
axes[0].set_xlabel("Study Time")
axes[0].set_ylabel("G1 Grade")

# Plot 2: Study time vs G2
axes[1].scatter(df['studytime'], df['G2'], color='orange', alpha=0.5)
axes[1].set_title("Distribution of Second period grade (G2)")
axes[1].set_xlabel("Study Time")
axes[1].set_ylabel("G2 Grade")

# Plot 3: Study time vs G3
axes[2].scatter(df['studytime'], df['G3'], color='green', alpha=0.5)
axes[2].set_title("Distribution of Final Grades (G3)")
axes[2].set_xlabel("Study Time")
axes[2].set_ylabel("G3 Grade")

plt.tight_layout()
plt.show()
```

* So Graphs also saying that there is no any correlation between study time & Grades

## Bar chart: male vs female average score.

```python
plt.figure(figsize=(6, 6))

# Calculate averages
female_avg = df[df['sex'] == 'F']['marks'].mean()
male_avg = df[df['sex'] == 'M']['marks'].mean()

# Create a bar chart for the averages
genders = ['Female', 'Male']
averages = [female_avg, male_avg]
bar_colors = ['blue', 'orange']

plt.bar(genders, averages, color=bar_colors, width=0.5, alpha=0.7)

plt.title("Average Marks Based on Gender")
plt.xlabel("Gender")
plt.ylabel("Average Marks")

# Add the numeric values on top of the bars for clarity
for i, v in enumerate(averages):
    plt.text(i, v + 0.5, f'{v:.2f}', ha='center', fontweight='bold')

plt.ylim(0, max(averages) * 1.2) # Give a little headroom for the text
plt.show()
```
